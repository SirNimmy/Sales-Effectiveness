## ANALYSIS:

### Sales Effectiveness:
    -Sales effectiveness is all about how well a company’s sales team can close deals at every step of the customer’s buying journey—and do it at the right time and on the right terms. But improving sales effectiveness isn’t just the sales team’s job; it’s a company-wide effort. Sales and marketing need to work closely together to figure out what’s working, what’s not, and how to keep improving. This means refining their knowledge, messaging, skills, and strategies to win more business.

    -In the past, "sales effectiveness" mainly referred to technology and consulting services designed to help businesses boost their sales results. Today, it’s a more holistic approach that focuses on continuous improvement and collaboration across teams.

### Sales Force Effectiveness:
    -The goal of sales force effectiveness is simple: drive more revenue by bringing in new customers, increasing product or service sales, and encouraging existing customers to buy more through up-selling and cross-selling.

    -To track how well the sales team is performing, companies use sales force effectiveness metrics. These help measure both the overall sales team’s success and individual salesperson performance.
    
    -Looking at just total sales doesn’t always tell the full story. By comparing different performance metrics, businesses can get a clearer picture of each salesperson’s strengths and areas for improvement. On a larger scale, if revenue per salesperson is increasing, it’s a strong sign that the sales force is becoming more effective.

### Target variable == Status:
    -The target variable consists of **11 different labels** (*Junk Lead, Not Responding, CONVERTED, Just Enquiry, Potential, Long Term, In Progress Positive, In Progress Negative, LOST, Open, converted*), each representing a **customer lead category**.
    
    -These labels can be grouped based on their **potential**, classifying leads as either **high potential** or **low potential**.

### Feature Descriptions:

1. **Created:**
   - A **unique feature** that tracks activities related to **sales transactions**, including the **date and time** of goods sold.

2. **Product ID:**
   - A unique identifier assigned to each product.

3. **Source:**
   - This feature captures how customers **interact with the business**, including sources like **calls, live chats, and campaigns**.

4. **Mobile:**
   - A **unique feature** containing the **customer’s mobile number**.

5. **Email:**
   - Another **unique feature** storing the **customer’s email ID**.

6. **Sales Agent:**
   - The **sales agent** serves as the **frontline representative** responsible for customer interactions.
   - Sales agents act on behalf of the company to introduce products to potential buyers and **earn commissions based on successful deals**.

7. **Location:**
   - Represents the **location of the sale**, typically set to the **main business address** but can be changed manually.
   - **Sales tax is calculated based on this location.**
   - This feature includes **various different locations**.

8. **Delivery Mode:**
   - Describes the **method of product delivery**, which can be one of three types:
     - **Actual Delivery:** Physical handover of goods.
     - **Symbolic Delivery:** A symbolic gesture (e.g., handing over documents) when goods are too large to move.
     - **Constructive Delivery:** No physical handover; the buyer is legally recognized as the new owner.

9. **Status (Target Variable):**
   - The **lead category of a customer**, classified into **high potential** and **low potential** leads.

## DATA SUMMARY:

### 2.K Neighbour Classifier

```python
from sklearn.neighbors import KNeighborsClassifier
error_rate = [] # Creating a empty list
for i in range(2,11):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train,y_train)
    y_knn_predict = knn.predict(X_test)
    error_rate.append(np.mean(y_knn_predict != y_test))
print("Error rate:",error_rate)
```

```python
# Step:1 [imutation of parameter]
from sklearn.model_selection import RandomizedSearchCV
# not used grid search cv beacuse memory reason

n_estimators = [int(x) for x in np.linspace(start=100, stop=2000, num=10)]
max_features = ['auto', 'sqrt'] ### auto - select randomly, sqrt - 
max_depth = [int(x) for x in np.linspace(10, 110, num=10)] ### level in the tree
max_depth.append(None)
min_samples_split = [2, 5, 15]
min_samples_leaf = [1, 2, 4]


# Step:2 make a dictionary of parameter
random_grid = {'n_estimators': n_estimators, 'max_features': max_features,
               'max_depth': max_depth, 'min_samples_split': min_samples_split,
               'min_samples_leaf': min_samples_leaf}

# Step:3 Object creation
rf_clf1 = RandomForestClassifier(random_state=42) 
# provide random state beacuse select row and column randomly


# Step:4 randomized search cv with parameter
rf_cv = RandomizedSearchCV(estimator=rf_clf1, scoring='f1',param_distributions=random_grid, n_iter=100, cv=3, 
                               verbose=3, random_state=42, n_jobs=-1)


# Step:5 fitting the training data
rf_cv.fit(X_train, y_train)

# Step:6 Getting the best parameter
rf_best_params = rf_cv.best_params_
print(f"Best paramters: {rf_best_params})")
```
